```python
#Part C: Data Preparation
11)le = LabelEncoder()
df['Gender']=le.fit_transform(df['Gender'])
df['SubscriptionType']=le.fit_transform(df['SubscriptionType'])
df['FavoriteGenre']=le.fit_transform(df['FavoriteGenre'])
df['SubscriptionRenewed']=le.fit_transform(df['SubscriptionRenewed'])
```

```python
12)y = df['SubscriptionRenewed']
x = df.drop(['UserID', 'SubscriptionRenewed'], axis=1)
```

```python
13)x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
#Part D: Decision Tree Classification
14)dt = DecisionTreeClassifier(random_state=42)
dt.fit(x_train,y_train)
pred = dt.predict(x_test)
```

```python
15)print('Decision Tree Accuracy:',accuracy_score(y_test,pred))
```

```python
16)print('\nConfusion Matrix')
print(confusion_matrix(y_test,pred))
```

```python
#Part E: K-Nearest Neighbors (KNN)
17)knn = KNeighborsClassifier(n_neighbors=5)
knn.fit(x_train,y_train)
knn_pred = knn.predict(x_test)
```

```python
18)print('KNN Accuracy:',accuracy_score(y_test,knn_pred))
print('Decision Tree Accuracy:',accuracy_score(y_test,pred))
```
