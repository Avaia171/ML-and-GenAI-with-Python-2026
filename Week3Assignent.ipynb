```python
from sklearn.preprocessing import LabelEncoder
```

```python
le = LabelEncoder()
df['crop_type']=le.fit_transform(df['crop_type'])
df['soil_type']=le.fit_transform(df['soil_type'])
df.head()
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
x = df.drop('yield_ton_per_hectare', axis=1)
y = df['yield_ton_per_hectare']
#yield_per_hectare as target variable
x_train_reg,x_test_reg,y_train_reg,y_test_reg = train_test_split(x,y,test_size=0.2,random_state=42)
lr = LinearRegression()
print("x_train shape:", x_train_reg.shape)
print("x_test shape:", x_test_reg.shape)
print("y_train shape:", y_train_reg.shape)
print("y_test shape:", y_test_reg.shape)
```

```python
lr = LinearRegression()
lr.fit(x_train_reg, y_train_reg)
print("Coefficients:", lr.coef_)
print("Intercept:", lr.intercept_)
```
